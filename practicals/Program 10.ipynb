```python
import nltk
from transformers import pipeline

print("========== EXPERIMENT 10 / MINI-PROJECT: TRANSFORMER-BASED NLP PIPELINE ==========")

# 1. Text Summarization Pipeline
summarizer = pipeline("summarization", model="sshleifer/distilbart-cnn-12-6")

article = """
Natural Language Processing (NLP) is a crucial subfield of Artificial Intelligence that enables computers to understand, 
interpret, and generate human language. Modern NLP leverages deep neural networks, particularly Transformer architectures 
like BERT, GPT, and T5, to achieve state-of-the-art performance across machine translation, text summarization, 
and conversational agents. By utilizing self-attention mechanisms, Transformers model long-range contextual relationships 
effectively without recurrence.
"""

print("\n--- Original Text ---")
print(article.strip())

print("\n--- Generating Abstractive Summary ---")
summary = summarizer(article, max_length=50, min_length=20, do_sample=False)
print("Summary:", summary[0]['summary_text'])

# 2. Sentiment Analysis Pipeline
sentiment_analyzer = pipeline("sentiment-analysis", model="distilbert-base-uncased-finetuned-sst-2-english")

test_sentences = [
    "This NLP laboratory assignment provides comprehensive hands-on machine learning experience!",
    "The out-of-vocabulary rate was too high and caused performance degradation."
]

print("\n--- Sentiment Analysis Evaluation ---")
for sent in test_sentences:
    res = sentiment_analyzer(sent)[0]
    print(f"Sentence: '{sent}'")
    print(f"Sentiment: {res['label']} (Confidence: {res['score']:.4f})\n")

print("========== EXPERIMENT 10 COMPLETED SUCCESSFULLY ==========")
```
